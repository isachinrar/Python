```python
#using the scatter plot

import matplotlib.pyplot as plt

study_hours = [1,2,4,1,5,6,9]
exam_score = [45,66,77,55,66,78,90]


plt.scatter(study_hours,exam_score)
plt.title('study_hours vs exam_houre')
plt.colorbar()
plt.xlabel('study_hours')
plt.ylabel('exam_hours')
plt.grid(True)
plt.show()
```

```python
# Size of points based on score (bigger score -> bigger point)
sizes = [score * 2 for score in exam_score]
colors = ['red' if score < 60 else 'green' for score in exam_score]

plt.scatter(study_hours, exam_score, s=sizes, c=colors)
plt.title('Colored & Sized Scatter Plot')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.grid(True)
plt.show()
```

```python
#multiple graph in one table 
# Assume two groups: Class A and Class B
class_a_hours = [2, 4, 6, 8]
class_a_scores = [45, 55, 65, 85]

class_b_hours = [1, 3, 5, 7, 9]
class_b_scores = [40, 50, 60, 70, 90]

plt.scatter(class_a_hours, class_a_scores, label='Class A', color='blue')
plt.scatter(class_b_hours, class_b_scores, label='Class B', color='orange')

plt.title('Scatter Plot: Class A vs Class B')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.legend()
plt.grid(True)
plt.show()
```
