```python
#use the pic chart
import matplotlib.pyplot as plt
labels = ["sachin","kolihi","rohit"]
runs = [18000,2000,10000]

plt.title("career run of the indian batmans")
plt.pie(runs,labels=labels)
plt.show
```

```python
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.pie(runs, labels=labels, colors=colors)
plt.show()
```

```python
#use tthe add edges and style
#wedgeprops
# Pie chart with style (wedgeprops)
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    wedgeprops={'edgecolor': 'black', 'linewidth': 2, 'linestyle': '--'}
)
plt.show()
```

```python
#expload and highlight one slice

explode = [0.1, 0, 0]  # Only highlight Sachin's slice

plt.pie(runs, labels=labels, explode=explode, colors=colors)
plt.title("Exploded Pie Example")
plt.show()
```

```python
#make more attractive of the highlightes slices

plt.pie(
    runs,
    labels=labels,
    explode=explode,
    colors=colors,
    shadow=True  # adds a 3D-like shadow
)
plt.show()
```

```python
#start angle use the roted the slice

plt.pie(
    runs,
    labels=labels,
    explode=explode,
    colors=colors,
    shadow=True,
    startangle=140
)
plt.show()
```

```python
plt.pie(
    runs,
    labels=labels,
    autopct='%1.1f%%',
    startangle=140
)
plt.title("Career Run Share")
plt.show()
```
