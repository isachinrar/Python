```python
import pandas as pd
#sample data frame 

data = {
    'Name': ['Alic','Sachin','Vicky'],
    'Math': [34,46,67],
    'Science': [77,88,99],
    'English': [98,78,77]

}

df = pd.DataFrame(data)

df
```

```python
#using the melt 
df.melt(id_vars=["Name"], value_vars=["Math", "Science", "English"], var_name="Subject", value_name="Score")
```

```python
#pivot() — Long to Wide

df_melted = df.melt(id_vars=["Name"], value_vars=["Math","Science","English"],var_name="Subject",value_name="Score")
df_melted

#df_melted.pivot_table(index="Name", columns="Subject", values="Score")
```
